```python
history_dir = "histories"
plot_dir = "plots/history"

history_files = sorted(
    glob.glob(os.path.join(history_dir, "*_history.csv"))
)

print(f"Found {len(history_files)} history files.")

for history_file in history_files:

    filename = os.path.basename(history_file)

    model_name = filename.replace("_history.csv", "")

    history = pd.read_csv(history_file)

    required_columns = [
        "loss",
        "accuracy",
        "val_loss",
        "val_accuracy"
    ]

    missing_columns = [
        column
        for column in required_columns
        if column not in history.columns
    ]

    if missing_columns:
        print(
            f"Skipping {model_name}: "
            f"missing columns {missing_columns}"
        )
        continue

    epoch = range(1, len(history) + 1)

    fig, axes = plt.subplots(
        2,
        1,
        figsize=(8, 7),
        sharex=True
    )

    axes[0].plot(
        epoch,
        history["loss"],
        label="Training"
    )

    axes[0].plot(
        epoch,
        history["val_loss"],
        label="Validation"
    )

    axes[0].set_ylabel("Loss")
    axes[0].set_title("(a) Loss")
    axes[0].legend()
    axes[0].grid(axis="y", alpha=0.3)

    axes[1].plot(
        epoch,
        history["accuracy"],
        label="Training"
    )

    axes[1].plot(
        epoch,
        history["val_accuracy"],
        label="Validation"
    )

    axes[1].set_xlabel("Epoch")
    axes[1].set_ylabel("Accuracy")
    axes[1].set_title("(b) Accuracy")
    axes[1].set_ylim(0, 1)
    axes[1].legend()
    axes[1].grid(
        axis="y",
        alpha=0.3
    )

    title = display_names.get(
        model_name,
        model_name.replace("_", " ")
    )
    
    fig.suptitle(title)

    fig.tight_layout()

    fig.subplots_adjust(
        top=0.90
    )

    output_file = os.path.join(
        plot_dir,
        f"{model_name}_history.png"
    )

    fig.savefig(
        output_file,
        dpi=300,
        bbox_inches="tight"
    )

    plt.close(fig)

    print(f"Saved: {output_file}")


print("\nHistory plotting complete.")
```

```python
confusion_matrix_dir = "confusion_matrices"
plot_dir = "plots/confusion_matrices"

confusion_matrix_files = sorted(
    glob.glob(os.path.join(confusion_matrix_dir, "*.csv"))
)

print(
    f"Found {len(confusion_matrix_files)} "
    "confusion matrix files."
)

for matrix_file in confusion_matrix_files:

    filename = os.path.basename(matrix_file)

    model_name = os.path.splitext(filename)[0]

    cm_df = pd.read_csv(matrix_file)

    cm_df = cm_df.apply(
        pd.to_numeric,
        errors="coerce"
    )

    if cm_df.shape != (7, 7):
        print(
            f"Skipping {filename}: "
            f"expected a 7x7 matrix, found {cm_df.shape}."
        )

        continue

    #try:
        #cm_df = cm_df.loc[
            #EMOTION_LABELS,
            #EMOTION_LABELS
        #]

    #except KeyError:
        #print(
            #f"Skipping {filename}: "
            #"emotion labels do not match the expected order."
        #)

        #continue

    cm = cm_df.to_numpy()

    fig, ax = plt.subplots(figsize=(8, 7))

    image = ax.imshow(
        cm,
        vmin=0,
        vmax=1
    )

    ax.set_xticks(range(len(EMOTION_LABELS)))

    ax.set_yticks(range(len(EMOTION_LABELS)))

    ax.set_xticklabels(
        EMOTION_LABELS,
        rotation=45,
        ha="right"
    )

    ax.set_yticklabels(EMOTION_LABELS)

    ax.set_xlabel("Predicted emotion")

    ax.set_ylabel("Actual emotion")

    for i in range(7):
        for j in range(7):
            value = cm[i, j]

            text_colour = (
                "white"
                if value >= 0.5
                else "black"
            )

            ax.text(
                j,
                i,
                f"{value:.1%}",
                ha="center",
                va="center",
                color=text_colour
            )

    colourbar = fig.colorbar(image, ax=ax)

    colourbar.ax.yaxis.set_major_formatter(PercentFormatter(1.0))

    colourbar.set_label("Proportion of actual class")

    title = display_names.get(
        model_name,
        model_name.replace("_", " ")
    )

    ax.set_title(title)

    fig.tight_layout()

    output_file = os.path.join(
        plot_dir,
        f"{model_name}_confusion_matrix.png"
    )

    fig.savefig(
        output_file,
        dpi=300,
        bbox_inches="tight"
    )

    plt.close(fig)

    print(f"Saved: {output_file}")


print("\nConfusion matrix plotting complete.")
```
